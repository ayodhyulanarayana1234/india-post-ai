```python
# Convert coordinates to numeric
df["latitude_num"] = pd.to_numeric(df["latitude"], errors="coerce")
df["longitude_num"] = pd.to_numeric(df["longitude"], errors="coerce")

# Valid approximate geographic bounds for India
valid_lat = df["latitude_num"].between(6, 38)
valid_lon = df["longitude_num"].between(68, 98)

# 1. Normal coordinates
normal = valid_lat & valid_lon

# 2. Strong candidates for reversed coordinates
reversed_coords = (
    df["latitude_num"].between(68, 98) &
    df["longitude_num"].between(6, 38)
)

# 3. Missing or zero coordinates
missing_or_zero = (
    df["latitude_num"].isna() |
    df["longitude_num"].isna() |
    (df["latitude_num"] == 0) |
    (df["longitude_num"] == 0)
)

# 4. Other invalid/corrupted coordinates
invalid = ~(normal | reversed_coords | missing_or_zero)

# Results
print("Normal:", normal.sum())
print("Potentially reversed:", reversed_coords.sum())
print("Missing/zero:", missing_or_zero.sum())
print("Other invalid/corrupted:", invalid.sum())

print("\nTotal:", len(df))
print(
    "Check:",
    normal.sum()
    + reversed_coords.sum()
    + missing_or_zero.sum()
    + invalid.sum()
)
```
